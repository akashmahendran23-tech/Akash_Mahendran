```python
X = data.drop(columns="Outcome")
Y = data["Outcome"]
```

```python
print(X.shape)
print(Y.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.20, random_state=42)
```

```python
print(X_train.shape)
print(X_test.shape)
print(Y_train.shape)
print(Y_test.shape)
```

## Feature engineering

```python
X_train['Insulin'] = np.log1p(X_train['Insulin'])
X_test['Insulin'] = np.log1p(X_test['Insulin'])
```

```python
X_train['Pregnancies'] = np.log1p(X_train['Pregnancies'])
X_test['Pregnancies'] = np.log1p(X_test['Pregnancies'])
```

```python
print(skew(X_train['Insulin']))
plt.figure(figsize=(8,5))
sns.histplot(data=X_train['Insulin'], color='pink', edgecolor = 'red')
print("Transformed distribution of Insulin")
```

## Data preprocessing

```python
from sklearn.preprocessing import MinMaxScaler

sc = MinMaxScaler()

sc
```

```python
X_train = sc.fit_transform(X_train)
X_test = sc.fit_transform(X_test)
```

```python
X_train
```

## Model Building

```python
from sklearn.linear_model import LogisticRegression

logrg = LogisticRegression(class_weight="balanced")
# logrg = LogisticRegression()
logrg
```

```python
logrg.fit(X_train, Y_train)
```

```python
Y_pred = logrg.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

acc = accuracy_score(Y_test, Y_pred)

acc
```

```python
data['Outcome'].value_counts()
```

```python
from sklearn.metrics import recall_score

recall_val = recall_score(Y_test, Y_pred)

recall_val
```

*recall_val is around 61%  
This says, out of all predicted cases as Diabetics, the model succedded to make 61% correct predicttion as Diabetic,  
means, incorrectly marked 39% values as Diabetics*

```python
from sklearn.metrics import precision_score

prec_val = precision_score(Y_test, Y_pred)

prec_val
```
